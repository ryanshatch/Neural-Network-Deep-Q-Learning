```python
from tensorflow.keras import backend as K   # Import Keras backend to clear the session so that the memory is freed up after training the model 
                                            # That ways I can train the model (Ex. Running the Examples 1 - 4) redundantly without having to restart the kernel
K.clear_session()                           # Clear the session
```

```python
# * Improving the simple net in Keras with hidden layers:

# We have a baseline accuracy of 92.36% on training, 92.27% on validation, and 92.22% on the
# test. This is a good starting point, but we can certainly improve it. Let's see how.
# A first improvement is to add additional layers to our network. So, after the input layer, we
# have a first dense layer with the N_HIDDEN neurons and an activation function relu. This
# additional layer is considered hidden because it is not directly connected to either the input
# or the output. After the first hidden layer, we have a second hidden layer, again with
# the N_HIDDEN neurons, followed by an output layer with 10 neurons, each of which will fire
# when the relative digit is recognized. The following code defines this new network:

import numpy as np  # Import NumPy for numerical computing
import tensorflow as tf # Import TensorFlow for deep learning
from tensorflow.keras.datasets import mnist # Import the MNIST dataset
from tensorflow.keras.models import Sequential  # Import the Sequential model from Keras
from tensorflow.keras.layers import Dense, Activation   # Import the Dense and Activation layers from Keras
from tensorflow.keras.optimizers import SGD # Import the SGD optimizer from Keras
from tensorflow.keras.utils import to_categorical   # Import to_categorical from Keras

np.random.seed(1671)  # The seed for the random number generator is to make sure that the results are able to be reproduced

# Parameters for the network and training data
NB_EPOCH = 20
BATCH_SIZE = 128
VERBOSE = 1
NB_CLASSES = 10 # Set the number of classes to 10 because there are 10 digits (0-9)
OPTIMIZER = SGD()  # Using the Stochastic Gradient Descent optimizer for the network training process
N_HIDDEN = 128 # Number of neurons in the hidden layer of the network so that it can learn the features that are important for the classification
VALIDATION_SPLIT = 0.2 # 20% of the training data is used for validation during the training process

# Data: shuffled and split between train and test sets
(X_train, y_train), (X_test, y_test) = mnist.load_data()    # Load the MNIST dataset and split it into training and testing data sets
RESHAPED = 784  # Reshaping the data to a vector based off of the number of features in the input data. 
                #* Since each image is 28x28px I am reshaping the data to a vector of 784 elements

# Normalize the data to be between 0 and 1
X_train = X_train.reshape(60000, RESHAPED).astype('float32') / 255 # Reshape the training data to be a vector of 784 elements and normalize it to be between 0 and 1
X_test = X_test.reshape(10000, RESHAPED).astype('float32') / 255   # Reshape the testing data to be a vector of 784 elements and normalize it to be between 0 and 1

print(X_train.shape[0], 'train samples')    # Print the number of training samples in the dataset
print(X_test.shape[0], 'test samples')      # Print the number of testing samples in the dataset

# Represent the targets as one-hot-encoded vectors and convert the labels into a set of 10 numbers to input into the network
Y_train = to_categorical(y_train, NB_CLASSES)   # Convert the training labels into a set of 10 numbers to input into the network
Y_test = to_categorical(y_test, NB_CLASSES)    # Convert the testing labels into a set of 10 numbers to input into the network

#* Defining the Model and network with two hidden layers:
model = Sequential()                # Define the model as a Sequential model in Keras
model.add(Dense(N_HIDDEN, input_shape=(RESHAPED,))) # Add a Dense layer with N_HIDDEN neurons and the input shape of the data
model.add(Activation('relu'))       # Add an activation function to the layer to introduce non-linearity to the model
model.add(Dense(N_HIDDEN))          # Add another Dense layer with N_HIDDEN neurons to the model so that it can learn more features
model.add(Activation('relu'))       # Add another activation function to the layer to introduce non-linearity to the model
model.add(Dense(NB_CLASSES))        # Add the output layer with NB_CLASSES neurons to the model so that it can classify the input data
model.add(Activation('softmax'))    # Add the softmax activation function to the output layer to get the probabilities of each class

#* Print a summary of the model to see the architecture of the model / network along with the number of the parameters that were used to train each Layer.
model.summary() 

# Compile the model with the loss function, optimizer, and metrics to evaluate the model during its training and testing
model.compile(loss='categorical_crossentropy',  # Use the categorical cross-entropy loss function to train the model
              optimizer=OPTIMIZER,           # Use the Stochastic Gradient Descent optimizer to train the model
              metrics=['accuracy'])         # Use the accuracy metric to evaluate the model during training and testing

# Train the model with the training data and evaluate it with the testing data
history = model.fit(X_train, Y_train,   # Train the model with the training data and the labels for the training data
                    batch_size=BATCH_SIZE, epochs=NB_EPOCH, # Set the batch size and number of epochs for the training process
                    verbose=VERBOSE, validation_split=VALIDATION_SPLIT) # Set the verbosity and validation split for the training process to evaluate the model during training

# Evaluate the model with the testing data to see how well it generalizes to unseen data and print the test's score and the models overall accuracy
score = model.evaluate(X_test, Y_test, verbose=VERBOSE) # Evaluate the model with the testing data to see how well it generalizes to unseen data
print("Test score:", score[0])  # Print the test score of the model 
print('Test accuracy:', score[1])   # Print the test accuracy of the model to be able to see and compare the training and testing accuracy between each model.
```
